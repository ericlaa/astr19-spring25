```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1000 random values using beta distribution
data = np.random.beta(a=2.0, b=5.0, size=1000)
```

```python
# Plot histogram
plt.hist(data, bins=100, color='salmon', edgecolor='black')
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Beta Distributed Random Numbers (a=2, b=5) Histogram")
plt.savefig("p10_betahist.pdf")
plt.show()
```
