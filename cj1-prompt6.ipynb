This notebook contains code for a program which uses numpy to compute the sine and cosine of a function, generate an evenly spaced list of 1000 values from 0 to 2π, and finally print a table of the first 10 values and their computed result, using our sine and cosine functions.

## Importing numpy
The line below imports the numpy library which contains many mathematical functions for Python. 
We will later use the library to compute the sine and cosine of a function, and generate evenly spaced values.

```python
import numpy as np
```

## Sine function

The sin_x(x) function below computes and returns the sine of x, using the numpy library.

```python
def sin_x(x):
    """
    Returns the sine of x using numpy's sin function.
    """
    return np.sin(x)
```

## Cosine function

The cos_x(x) function below computes and returns the cosine of x, using the numpy library.

```python
def cos_x(x):
    """
    Returns the cosine of x using numpy's cos function.
    """
    return np.cos(x)
```

## Evenly spaced value generation

The line below generates 1000 evenly spaced values of `x` between 0 and 2π using numpy's `linspace` function and stores them in a list.

```python
x_vals = np.linspace(0, 2 * np.pi, 1000)
```

## Printing the table

The line below prints out a pretty table heading, and then loops through the first 10 values from the values in the x_vals list.

Each iteration/row prints the original value, as well as the sine and cosine of those values from our functions `sin_x(x)` and `cos_x(x)`, respectively.

```python
print("x\t\tsin(x)\t\tcos(x)")
print("-" * 40)
for i in range(10):
    print(f"{x_vals[i]:.6f}\t{sin_x(x_vals[i]):.6f}\t{cos_x(x_vals[i]):.6f}")
```
