```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 100 evenly spaced values ranging from 0 to 1
x = np.linspace(0, 1, 100)
```

```python
# Define functions that use numpy sin/cos
def sin_func(x):
    return np.sin(x)

def cos_func(x):
    return np.cos(x)
```

```python
# Create a multipanel plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Plot sin(x)
ax1.plot(x, sin_func(x))
ax1.set_title("sin(x)")
ax1.set_xlabel("x")
ax1.set_ylabel("sin(x)")
ax1.grid(True)

# Plot cos(x)
ax2.plot(x, cos_func(x))
ax2.set_title("cos(x)")
ax2.set_xlabel("x")
ax2.set_ylabel("cos(x)")
ax2.grid(True)

# Save as a PDF
plt.savefig("p8_sincos.pdf")
plt.show()
```
