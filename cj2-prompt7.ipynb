```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 100 evenly spaced values ranging from 0 to 1
x = np.linspace(0, 1, 100)
```

```python
# A function that computes an exponential using numpy
def exponential(x):
    return np.exp(x)
```

```python
# Computes the exponential of x
y = exponential(x)
```

```python
# Plot awesomeness vs. time, saved as a PDF
plt.plot(x, y)
plt.xlabel("Time [milliseconds]")
plt.ylabel("Awesomeness")
plt.title("Awesomeness Over Time")
plt.grid(True)
plt.savefig("p7_awesomeness.pdf")
plt.show()
```
